# Check Invalid Values

```python
print('key -1 count:', (df['key'] == -1).sum())
print('time_signature invalid:', df[df['time_signature'] < 3]['time_signature'].value_counts())
print('duration zero count:', (df['duration_ms'] == 0).sum())
print('tempo zero count:', (df['tempo'] == 0).sum())
```

# Check Outliers in Duration

```python
print('duration stats:')
print(df['duration_ms'].describe())
print('\nsongs longer than 10 minutes:', (df['duration_ms'] > 600000).sum())
print('songs shorter than 30 seconds:', (df['duration_ms'] < 30000).sum())
```

# Check Outliers Using Domain Knowledge

```python
bounds = {
    'popularity':        (0, 100),
    'duration_ms':       (30000, 600000),
    'danceability':      (0, 1),
    'energy':            (0, 1),
    'loudness':          (-60, 5),
    'speechiness':       (0, 1),
    'acousticness':      (0, 1),
    'instrumentalness':  (0, 1),
    'liveness':          (0, 1),
    'valence':           (0, 1),
    'tempo':             (40, 250),
}

for col, (lower, upper) in bounds.items():
    outliers = df[(df[col] < lower) | (df[col] > upper)].shape[0]
    print(f'{col}: outliers={outliers}')
```

# Multiple Artists Analysis

```python
df['artist_count'] = df['artists'].str.split(';').str.len()
print(df['artist_count'].value_counts().sort_index())
```

# Investigate Songs with Too Many Artists

```python
print(df[df['artist_count'] >= 15][['track_name', 'artists', 'track_genre', 'artist_count']].to_string())
```

```python
df.drop(columns=['artist_count'], inplace=True)
```

# Duration Distribution in Minutes

```python
df['duration_min'] = df['duration_ms'] / 60000
print(df['duration_min'].describe())
print('\nsongs under 1 min:', (df['duration_min'] < 1).sum())
print('songs 1-3 min:', ((df['duration_min'] >= 1) & (df['duration_min'] <= 3)).sum())
print('songs 3-5 min:', ((df['duration_min'] > 3) & (df['duration_min'] <= 5)).sum())
print('songs 5-10 min:', ((df['duration_min'] > 5) & (df['duration_min'] <= 10)).sum())
print('songs over 10 min:', (df['duration_min'] > 10).sum())
df.drop(columns=['duration_min'], inplace=True)
```

# Songs with Same Name Different Artists

```python
duplicate_names = df[df['track_name'].duplicated(keep=False)]
print('rows with duplicate track names:', len(duplicate_names))
print('unique track names duplicated:', duplicate_names['track_name'].nunique())
print('\nexample:')
duplicate_names[['track_name', 'artists', 'track_genre']].head(10)
```

# Investigate True Duplicates (Same Track ID + Same Genre)

```python
true_duplicates = df[df.duplicated(subset=['track_id', 'track_genre'], keep=False)]
print('rows with true duplicates:', len(true_duplicates))
print('unique songs affected:', true_duplicates['track_id'].nunique())
print('\nexample:')
true_duplicates[['track_id', 'track_name', 'artists', 'track_genre']].sort_values('track_id').head(10)
```

```python
true_dup_counts = df.groupby(['track_id', 'track_genre']).size()
print(true_dup_counts[true_dup_counts > 2])
```

# Multi Genre Songs Count

```python
genre_counts = df.groupby('track_id')['track_genre'].count()
print('songs appearing in 1 genre:', (genre_counts == 1).sum())
print('songs appearing in 2 genres:', (genre_counts == 2).sum())
print('songs appearing in 3 genres:', (genre_counts == 3).sum())
print('songs appearing in 4+ genres:', (genre_counts >= 4).sum())
print('\ntotal unique songs:', genre_counts.shape[0])
```

```python
text_cols = ['track_name', 'artists', 'album_name', 'track_genre']

for col in text_cols:
    print(f'\n--- {col} ---')
    print('sample values:')
    print(df[col].head(5).tolist())
    print('has leading/trailing spaces:', df[col].str.startswith(' ').sum() + df[col].str.endswith(' ').sum())
    print('has special characters:', df[col].str.contains(r'[^\w\s\-\;\,\.\'\"\(\)\&\!\/\#\?]', regex=True, na=False).sum())
```

# Investigate Artists with Leading/Trailing Spaces

```python
mask = df['artists'].str.startswith(' ') | df['artists'].str.endswith(' ')
print(df[mask]['artists'].tolist())
```

# Investigate Special Characters in Text Columns

```python
for col in ['track_name', 'artists', 'album_name']:
    special = df[df[col].str.contains(r'[^\w\s\-\;\,\.\'\"\(\)\&\!\/\#\?\+\=\@\$\%\*]', regex=True, na=False)]
    print(f'\n--- {col} ---')
    print(f'count: {len(special)}')
    print(special[col].head(5).tolist())
```
